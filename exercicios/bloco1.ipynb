```python
import pandas as pd 
import numpy as np 

df = pd.read_excel('dados/economi.xlsx')

display(df)
display(df.head())
display(list(df.columns)) #descreve o nome das colunas em uma lista 
display(df.info())
display(df.describe())
```

```python
#seleção de colunas
display(df[['Ano', 'M - oferta monetaria']].head())

# tipos de dados
df['Ano'] = pd.to_datetime(df['Ano'], format='%Y')

display(df.info())


#Calculo de matemática entre colunas
df['P - nivel de precos'] = df['M - oferta monetaria'] * df['V - velocidade'] / df['Y - produto real']

display(df.head())
```

```python

# criar nova coluna

#calculo quantidade de moeda

df['Taxa de crescimento - M'] = df['M - oferta monetaria'].pct_change() * 100

#diferenca_m = df['Taxa de crescimento - M'] / 100

df['Media de crescimento - M'] = df['Taxa de crescimento - M'].mean() # média

#calculo velocidade de circulação
df['Taxa de crescimento - V'] = df['V - velocidade'].pct_change() * 100

# diferenca_v = df['Taxa de crescimento - V'] / 100

df['Media de crescimento - V'] = df['Taxa de crescimento - V'].mean() # média

#calculo produto

df['Taxa de crescimento - Y'] = df['Y - produto real'].pct_change() * 100



# diferenca_y = df['Taxa de crescimento - Y'] / 100

df['Media de crescimento - Y'] = df['Taxa de crescimento - Y'].mean() # média

#calculo preco

df['Taxa de crescimento - P'] = df['P - nivel de precos'].pct_change() * 100

#diferenca_p = df['Taxa de crescimento - P'] / 100

df['Media de crescimento - P'] = df['Taxa de crescimento - P'].mean() # média




# diferença  p= m + v - y

df['Inflacao aproximada'] = df['Taxa de crescimento - M'] + df['Taxa de crescimento - V'] - df['Taxa de crescimento - Y']

# df['Diferenca - P'] = np.where(df['P - real']==df['P - nivel de precos'], 0 , df['P - real']-df['P - nivel de precos'] ,  )

df['Erro aproximacao'] = df['Taxa de crescimento - P'] - df['Inflacao aproximada']

display(df.head())

df.info()
list(df.columns)
```

```python
# remover coluna

df.drop(columns=['P - real', 'Inflação aproximada', 'Diferenca - P' ], inplace=True)

list(df.columns)



```

```python
df.info()
```

```python
display(df)

# 1. Encontra o índice da linha com o maior valor
maior_inflacao = df['Inflacao aproximada'].idxmax()

# 2. Seleciona a linha desse índice e apenas as colunas
resultado_inflacao = df.loc[[maior_inflacao], ['Trimestre', 'Inflacao aproximada']]

#ou, uma forma mais limpa, usando o nlargest, que retorna com a maior linha

maior_preco = df.nlargest(1,'Taxa de crescimento - P')[['Trimestre', 'Taxa de crescimento - P']]

# seleciona uma unica linha com a media de crescimento.

display(df['Media de crescimento - M'].iloc[0])

media = df['Taxa de crescimento - M'].mean()
display(media)

# comparacao para descobrir periodos em que a velocidade diminuiu

diferenca_velocidade = df['V - velocidade'].diff()

menor_velocidade = df[diferenca_velocidade < 0]

display(menor_velocidade[['Trimestre', 'V - velocidade', 'Taxa de crescimento - V']])



display(maior_preco)
display(resultado_inflacao.style.set_caption("Maior inflação registrada"))

#Qual foi a diferença entre a inflação aproximada e a inflação calculada pela variação de P?

display(df[['Taxa de crescimento - P','Inflacao aproximada', 'Erro aproximacao']])


#O crescimento da oferta monetária foi sempre acompanhado de inflação positiva?

df['Inflacao positiva?'] = np.where(df['Inflacao aproximada'] > 0, 'Sim', 'Não')

df['crescimento m positivo'] = np.where(df['Taxa de crescimento - M'] > 0, 'Sim', 'Não')

#verifica se os resultados correspondem
corresponde_m_i = df['crescimento m positivo'] == df['Inflacao positiva?']

resultado_m_i = df[corresponde_m_i]

display(resultado_m_i[['Trimestre', 'Inflacao positiva?', 'crescimento m positivo', 'Inflacao aproximada', 'Taxa de crescimento - M' ]])

#respondendo a pregunta original

excessoes = df[
    (df['Taxa de crescimento - M'] > 0)
    & (df['Taxa de crescimento - P'] <= 0)
]

display(excessoes[[
    'Trimestre',
    'Taxa de crescimento - M',
    'Taxa de crescimento - P'
    
]])

# display(df['Inflacao positiva?'])



```
