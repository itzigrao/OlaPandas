```python
import numpy as np

M = np.array([1000, 1200, 1500])
V = np.array([2.0, 1.8, 1.5])
Y = np.array([500, 550, 600])

P = M * V / Y

simulacao_crescimento_m = M * 1.10 

p_novo = simulacao_crescimento_m * V / Y

variacao_p = (p_novo/P - 1) * 100

simulacao = np.column_stack([simulacao_crescimento_m, p_novo, variacao_p])
# cada linha representa um cenário, e cada coluna representa uma variável.

display(simulacao)
display(simulacao.shape)
display(simulacao.ndim)
display(simulacao.dtype)
```

```python
#axis=0 coluna
#axis=1 linha

media_coluna = simulacao.mean(axis=0)
media_linha = simulacao.mean(axis=1)

display(simulacao.shape)

# np.round serve para arredondar variavei para uma determinada quantidade de casas decimais
arredondado = np.round(media_linha, decimals=2)

display(arredondado)
```

```python
#desvio padrao

serie_a = np.array([10,10,10,10])
serie_b = np.array([5, 10, 10, 15])

desvio_padrao_a = np.std(serie_a) #0.0
desvio_padrao_b = np.std(serie_b) #3.53

print(desvio_padrao_a)
print(desvio_padrao_b)

# sintaxe básica: np.std(a, axis=None, ddof=0) lembrando que axis define a dimentsao, linha ou coluna. O A se refere ao dado
# ddof: ajuste dos graus de liberdade no denominador.


```

```python
# Exercício utilizando a array de simulação

#dimensões da matriz
display(simulacao.shape)

# media por coluna
media = np.mean(simulacao, axis=0)

#desvio padrao por coluna
desvio = np.std(simulacao, axis=0)

# arredondamento

media_arredondada = np.round(media, decimals=2)
desvio_arredondado = np.round(desvio, decimals=2)
```
