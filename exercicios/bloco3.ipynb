```python
import numpy as np

M = np.array([1000, 1200, 1500])
V = np.array([2.0, 1.8, 1.5])
Y = np.array([500, 550, 600])

P = M * V / Y

simulacao_crescimento_m = M * 1.10 

p_novo = simulacao_crescimento_m * V / Y

variacao_p = (p_novo/P - 1) * 100

simulacao = np.column_stack([simulacao_crescimento_m, p_novo, variacao_p])
# cada linha representa um cenário, e cada coluna representa uma variável.

display(simulacao)
display(simulacao.shape)
display(simulacao.ndim)
display(simulacao.dtype)
```

```python
#axis=0 coluna
#axis=1 linha

media_coluna = simulacao.mean(axis=0)
media_linha = simulacao.mean(axis=1)

display(simulacao.shape)

# np.round serve para arredondar variavei para uma determinada quantidade de casas decimais
arredondado = np.round(media_linha, decimals=2)

display(arredondado)
```

```python
#desvio padrao

serie_a = np.array([10,10,10,10])
serie_b = np.array([5, 10, 10, 15])

desvio_padrao_a = np.std(serie_a) #0.0
desvio_padrao_b = np.std(serie_b) #3.53

print(desvio_padrao_a)
print(desvio_padrao_b)

# sintaxe básica: np.std(a, axis=None, ddof=0) lembrando que axis define a dimentsao, linha ou coluna. O A se refere ao dado
# ddof: ajuste dos graus de liberdade no denominador.


```

```python
# broadcasting:

# Broadcasting serve para realizar operações entre arrays com dimensões diferentes, desde que sejam compatíveis.

precos = np.array([
    [100, 150, 300],
    [150, 250, 350]
])
display(precos.shape)


# agora, imagine que queremos apliccar um reajuste diferente à cada uma das colunas.

reajustes = np.array([1.10, 1.05, 1.20])

novos_precos = precos * reajustes
# O NumPy aplica os fatores correspondentes a cada coluna.


display(novos_precos.shape)

display(novos_precos)


media = np.mean(novos_precos, axis=0)

novos_precos = novos_precos.reshape(3,2) # 3 linhas 2 colunas

novos_precos = np.round(novos_precos, decimals=2)


display(novos_precos)

tudo = np.column_stack([novos_precos, media])

display(np.round(tudo, decimals=2))


```

```python
# aula 4
# indexação, slicing e filtros booleanos

# Vamos aprender a localizar, selecionar e filtrar informações dentro de arrays Numpy

# esses conhecimentos serão úteis em base de dados maiores.

# 1- Indexação

dados = np.array([
    [100, 200, 300],
    [400, 599, 600],
    [700, 800, 900]
])

display(dados.shape)

#linha inteira
print(dados[1])
# [400 500 600]

# coluna inteira
print(dados[:, 1])
# [200 500 800]


# 2- Slicing: 
# é uma técnica utilizada para selecionar sequências de elementos sem acessar cada posição individualmente.
# parametros necessarios inicio, fim, intervalo entre posições

valores = np.array([10, 20, 30, 40, 50, 60])

print(valores[1:4])
# [20 30 40]

print(valores[::2])
# [10 30 50]

#Para um array bidimensional, podemos utilizar dois intervalos

# 3- Mascaras booleanas: filtrando dados


inflacao = np.array([1.2, 2.5, 0.8, 3.4, 1.7, 4.1])


#Queremos selecionar apenas as observações com inflação superior a 2%.

condicao = inflacao > 2

display(condicao)
# booleanos

#Combinando condições

filtro = (inflacao > 2) & (inflacao < 4)

print(inflacao[filtro])
# com a mascara, aparece os valores presente no array
# [2.5 3.4]

# 4- np.where(): realizando operações condicionais
# np.where(condicao, valor_se_verdadeiro, valor_se_falso)

classificacao = np.where(
    inflacao > 2,
    "Acima de 2%",
    "Até 2%"
)

display(classificacao)
```

```python
# Exercício aula 4

moeda = np.array([1000, 1200, 1500])

politicas = np.array([0.02, 0.05, 0.10, 0.15])

politicas = politicas.reshape(4, 1)

resultado = moeda * (1 + politicas)

display(resultado.shape)

# 1. Selecione somente os resultados das políticas de expansão de 5% e 10%, utilizando slicing

resultado_1 = resultado[1:3,:] #linha e coluna. O : significa selecionar todas as posições da dimensão.
display(resultado_1)

# 2. Extraia todos os valores da terceira coluna, que representa o cenário econômico com oferta monetária inicial de 1.500.

resultado_2 = resultado[:, 2] # seleciono todas as linhas e somente a terceira coluna. dimensão ou formato do array seria = (4,)
display(resultado_2)

# 3. Utilize uma máscara booleana para selecionar apenas os resultados superiores a 1.300.

resultado_3 = resultado[resultado > 1300]
display(resultado_3)

# 4. Crie uma matriz de classificação usando np.where(): valores superiores a 1.500 recebem "Alto"; os demais recebem "Normal".

resultado_4 = np.where(resultado > 1500, 'Alto', 'Normal')
display(resultado_4)

# 5. Identifique as dimensões da matriz de classificação utilizando shape.

display(resultado_4.shape)
```

```python
# Aula 5: Funções matemáticas e operações cumulativas

#Agora vamos aprofundar cinco recursos do NumPy:

#np.sum(), np.min(), np.max(), np.cumsum() e np.cumprod().

# 1. Funções de agregação

# Agregação é o processo de combinar vários valores para produzir um indicador resumido.

receitas = np.array([
    [100, 150, 200],
    [120, 180, 220],
    [140, 160, 250],
    [160, 210, 270]
])

#Cada linha representa um trimestre e cada coluna representa um produto.

total = np.sum(receitas)
# Calcula a soma dos elementos dentro do array
display('Soma Total receita')
display(total)

receita_por_trimestre = np.sum(receitas, axis=1)
# soma os elementos de cada linha obtendo a receita total por trimestre
display('Receita por trimestre: ')
display(receita_por_trimestre)

receita_ano_produto = np.sum(receitas, axis=0)
# soma os elementos de cada coluna obtendo a receita total no ano por produto
display('Receita  anual total por produto')
display(receita_ano_produto)

menor_faturamento_produto = np.min(receita_ano_produto)
# identifica o menor numero(elemento) dentro da array
display('Valor do produto com menor faturameno anual:')
display(menor_faturamento_produto)

maior_faturamento_produto = np.max(receita_ano_produto)
# identifica o maior numero(elemento) dentro da array
display('Valor do produto com maior faturamento anual: ')
display(maior_faturamento_produto)

# ambos aceitam o parametro axis

# assim:
receita_trimestral_produto = np.max(receitas, axis=0)
display('Coluna contendo o maior faturamento de cada produto')
display(receita_trimestral_produto)

comparando = maior_faturamento_produto == receita_trimestral_produto
display(comparando)

```

```python
# 2. Operações cumulativas: np.cumsum()

receita = np.array([100, 120, 140, 160])

receita_acumulado = np.cumsum(receita)
#calcula uma sequencia de somas acumuladas enqaunto np.sum tras o total.

display('Evolução do faturamento acumulado')
display(receita_acumulado)

acumulado_produto = np.cumsum(receitas, axis=0) # acumulado das receitas por produto/coluna
display('acumulado das receitas por produto')
display(acumulado_produto)


acumulado_trimestre = np.cumsum(receitas, axis=1)
display('Acumulado das receitas por trimestre')
display(acumulado_trimestre)
```

```python
# 3. np.cumprod() — Produto acumulado
#  multiplica os valores sucessivamente, mantendo os produtos acumulados.

fatores = np.array([1.10, 1.05, 1.02])

resultado = np.cumprod(fatores)
display(resultado)

# A estrutura da chamada é: np.cumprod(array, axis=None)

```

```python
import numpy as np

# Exercício consolidação - Receitas e crescimento acumulado

receitas = np.array([
    [100, 150, 200], 
    [120, 180, 220],
    [140, 160, 250],
    [160, 210, 270]
])

#utilizando a matriz receita

#-- ANALISE COMERCIAL --

#faturamento anual de cada unidade 

faturamento_anual_unidade = np.sum(receitas, axis=0)
display(f'faturamento anual de cada unidade {faturamento_anual_unidade}')

#faturamento total por trimestre

faturamento_total_por_trimestre = np.sum(receitas, axis=1)
display(f'Faturamento total por trimestre {faturamento_total_por_trimestre}')

# maior receita trimestral por unidade

maior_receita_trimestral_por_unidade = np.max(receitas, axis=0)
display(f'Maior receita trimestral por unidade {maior_receita_trimestral_por_unidade}')

# faturamento acumulado por unidade ao logo dos 4 trimestres

faturamento_acumulado_unidade = np.cumsum(receitas, axis=0)
display(f'faturamnto acumulado unidade trimestre {faturamento_acumulado_unidade}')


# -- CRESCIMENTO ECONOMICO --

taxas = np.array([0.02, 0.03, -0.01, 0.04])

display(taxas.shape)

#adiciona um para calcular fatores de crescimento
fatores = 1 + taxas

crescimento_acumulado = np.cumprod(fatores)

variacao_acumulada = (crescimento_acumulado - 1) * 100 

display(crescimento_acumulado, variacao_acumulada)
```
