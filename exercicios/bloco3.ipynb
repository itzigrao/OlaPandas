```python
import numpy as np

M = np.array([1000, 1200, 1500])
V = np.array([2.0, 1.8, 1.5])
Y = np.array([500, 550, 600])

P = M * V / Y

simulacao_crescimento_m = M * 1.10 

p_novo = simulacao_crescimento_m * V / Y

variacao_p = (p_novo/P - 1) * 100

simulacao = np.column_stack([simulacao_crescimento_m, p_novo, variacao_p])

display(simulacao)
display(simulacao.shape)
display(simulacao.ndim)
display(simulacao.dtype)
```

```python
media_coluna = simulacao.mean(axis=0)
media_linha = simulacao.mean(axis=1)

display(simulacao.shape)

np.round(media_coluna)
```
