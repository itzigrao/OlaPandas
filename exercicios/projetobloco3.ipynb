```python
# Problema: uma empresa possui três unidades comerciais e deseja analisar seu desempenho em quatro
# trimestres, além de simular o impacto de diferentes políticas de crescimento sobre suas receitas.


import numpy as np

receitas = np.array([
    [120, 180, 220],
    [135, 175, 240],
    [150, 195, 230],
    [170, 210, 260]
])

taxas_projecao = np.array([0.03, 0.06, 0.10])



fatores = 1 + taxas_projecao

# 1. Diagnostico empresarial

faturamento_anual_unidade = np.sum(receitas, axis=0)
display('Faturamento anual por unidade: ', faturamento_anual_unidade)

faturamento_total = np.sum(receitas)
display('Faturamento total da empresa no período de um ano:', faturamento_total)

# Receita media trimestral de cada unidade

display(faturamento_anual_unidade.shape)
# 3 linhas

media_receita_trimestral = np.mean(receitas, axis=0)
display('Receita media por trimestre', media_receita_trimestral)


# Desvio padrao das receitas trimestrais por unidade

desvio_padrao_receitas_trimestre_unidade = np.std(receitas, axis=0)
display('Desvio padrão receitas trimestre por unidade', desvio_padrao_receitas_trimestre_unidade)
# -- PARTE UM OK --







```

```python
# 2. Analise temporal

# Para a variação percentual, utilize slicing para construir os vetores de receitas atuais e anteriores.


receita_por_trimestre = np.sum(receitas, axis=1)

faturamento_acumulado = np.cumsum(receita_por_trimestre)
display(' Faturamento acumulado anual', faturamento_acumulado)

display(receita_por_trimestre)
display(receita_por_trimestre.shape)
# 4 linhas

vetor_1 = receita_por_trimestre[:-1]
vetor_2 = receita_por_trimestre[1:]

display(vetor_1, vetor_2)

variacao = ((vetor_2 - vetor_1)/ vetor_1) * 100

display(variacao)
```

```python
# 3. Simulação de crescimento

display(taxas_projecao)

display(fatores)
display(fatores.shape)

fatores = fatores.reshape(3, 1)

display(faturamento_anual_unidade)
display(faturamento_anual_unidade.shape)

faturamento_anual_unidade = faturamento_anual_unidade.reshape(1,3) # uma linha
# eu devo aplicar todas as taxa de cresimento em cada resultado anual por produto

taxas_aplicadas = faturamento_anual_unidade * fatores

display('Projeções para cada produto: ', taxas_aplicadas)

```

```python
# 4. Classificção dos cenários

classificacao = np.where(taxas_aplicadas > 800, 'Alta', 'Regular')

extracao = faturamento_anual_unidade > 800

display(classificacao)
display(taxas_aplicadas[extracao])
```

```python
# 5. Identificar qual unidade apresenta maior faturamento anual, qual maior dispersao das receitas trimestrais e qual é o faturamento total da empresa projetado em cada cenario

maior_fatureamento_anual = faturamento_anual_unidade.max()
maior_dispersao = desvio_padrao_receitas_trimestre_unidade.max()


faturamento_total_cenarios = taxas_aplicadas.sum(axis=1)

display(maior_fatureamento_anual)
display(maior_dispersao)
display(faturamento_total_cenarios)
```
