```python
import pandas as pd 
import numpy as np 

df = pd.read_excel('dados/economi.xlsx')

display(df)
display(df.head())
display(list(df.columns)) #descreve o nome das colunas em uma lista 
display(df.info())
display(df.describe())
```

```python
#seleção de colunas
display(df[['Ano', 'M - oferta monetaria']].head())

# tipos de dados
df['Ano'] = pd.to_datetime(df['Ano'], format='%Y')

display(df.info())


#Calculo de matemática entre colunas
df['P - nivel de precos'] = df['M - oferta monetaria'] * df['V - velocidade'] / df['Y - produto real']

display(df.head())
```

```python
# criar nova coluna

df['Taxa de crescimento - M'] = df['M - oferta monetaria'].pct_change() * 100

diferenca_m = df['Taxa de crescimento - M'] / 100

df['Media de crescimento - M'] = diferenca_m.mean() # média

display(df)
```
